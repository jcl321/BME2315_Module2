## Disease Background: 
* Prevalence & incidence

FSHD has a prevalence of about 2.03 to 6.8 cases per 100,000 people worldwide, giving a rate of about 0.00203% to 0.0068%. In the U.S., prevalence is from 1 in 21,000 to 1 in 15,000, or a rate of 0.0048% to 0.0067% (Statland & Tawil, 2016). Estimates of incidence range from 1 in 14,826 to 1 in 7,500, giving rates of 0.0067% to 0.0133% (UMass Chan Medical School, 2014).

* Symptoms

Facial weakness, including difficulty to smile, whistle, or use a straw is an early sign of FSHD. Weakness in shoulders can also occur, leading to difficulty to reach with or lift the arms, and is evident through protrusion of the shoulder known as scapular winging. Leg drop occurs due to leg weakness and can lead to difficulty walking and tripping. Hip and abdominal weakness are also significant symptoms of FSHD (Muscular Dystrophy Association, 2015).

* Diagnosis

Diagnosis begins with a clinical examination that looks at patterns of muscle weakness that match with symptoms of FSHD. It is confirmed through genetic testing, where a repeat of the DZH4 gene on the 4th chromosome indicates possession of the disease (Muscular Dystrophy Association, 2015). Other parts of diagnosis may include blood tests to measure creatine kinase levels, neurological tests to look at patterns of muscle weakness, and muscle biopsies to analyze deterioration of muscle (FSHD Society, 2019). 

* Standard of care treatments (& reimbursement) 

No cures currently exist for FSHD. However, its progression can be slowed and symptom severity may be decreased with treatment. Monitoring potential  fatal complications with breathing, along with hearing and vision issues is common. Occupational physical therapy is also used to ease muscle pain, help range of motion, and help work against decline of muscle strength. Orthopedic surgeries may also be used to help treat injuries and prevent scapular winging (Tawil et al., 2015). Reimbursement from FSHD may come from private insurance claims and/or medicare. There is a significant increase in cost of care, totaling $19,400 each year in comparison to those without FSHD who average $5,250 in claims (Sasala, 2026).

* Disease Progression and Prognosis 

FSHD is a slowly progressive disease; however, there is high clinical variability in the rate and severity of progression. (Hamel et al., 2018) According to Hamel, earlier onset of FSHD symptoms of facial weakness is associated with greater disease severity as well as males are also reported to have greater severity of FSHD.  

As the disease continues to affect the body, the legs start to be affected and within 6 years of the onset of FSHD, there is a 24% chance of becoming wheelchair dependent. (Hamel) Also, according to Hamel, age increases the risk of needing a wheelchair. 

* Biological Mechanisms 

In healthy individuals, each person carries 11-100 repeat copies of D4Z4 retrogene, a transcription factor, on chromosome 4q35. (Hamel) Typically, this growth factor is not expressed in somatic tissues, however in individuals with 1-10 repeat copies of the retrogene have a contraction of the D4Z4 repeats and this results in hypomethylation and decreased repression of the heterochromatin. (Hamel) According to Hamel, because of this opening up of the chromatin, the D4Z4 genes are more accessible to be transcribed and translated into the protein DUX4. However, contraction and chromatin relaxation are not enough to cause FSHD as the entire strand of mRNA is not stable due to a lack of the polyadenylation signal; however, the 4qA haplotype contains a polymorphic adenylation signal what stabilizes the mRNA strand, allowing the expression of DUX4. (Hamel) According to Hamel, DUX4 is toxic to skeletal muscle and is normally only expressed in early development. When expressed, it also can result in cell death and inhibition of muscle regeneration. (Hamel)

Del-brax utilizes oligonucleotides and monoclonal antibodies to delivery siRNA to suppress DUX4 expression. EPI-321 is a one-time does epigenetic therapy that is still in the Phase I/II development, as of October 5, 2026, however, it has shown a statistically significant increases in whole-body lean muscle volume, favorable changes in biomarkers associated with DUX4 suppression, and favorable functional outcomes. ### Calculating Fat Fraction for Whole Region and TA Only

Each individual pixel contains a fat signal value and a water signal value. The code calculates the fat fraction by dividing the fat signal by the sum of the fat signal and the water signal, then writing it as a percentage. The fat fraction for the entire image region averages all of these amounts for each valid pixel, excluding pixels with both zero fat signal and zero water signal. For the TA muscle, this same process is used but only pixels included in the mask region are counted in the mean. Each of these outputs are placed into two arrays, fat_fraction_full, and fat_fraction_masked. The full array includes all valid pixels while the masked array only includes pixels within the TA region (ChatGPT).

```python
denominator = water + fat
valid_pixels = denominator > 0

fat_fraction = np.divide(
    fat,
    denominator,
    out=np.zeros_like(fat, dtype=float),
    where=valid_pixels
) * 100

# Keep the whole supplied image region for the first analysis.
fat_fraction_full = fat_fraction.copy()

# Then apply the TA mask for the muscle-specific analysis.
fat_fraction_masked = fat_fraction.copy()
fat_fraction_masked[~mask] = np.nan

whole_image_values = fat_fraction_full[valid_pixels]
muscle_values = fat_fraction_masked[~np.isnan(fat_fraction_masked)]
```

## Interpolation Analysis:

Fat fraction was measured across three different time points. First, a baseline measure was taken, then two subsequent measures were taken with a time interval of about one year between time points. The intervals are measured in months and provided for each patient. Since the time intervals were uneven, however, interpolation was used to allow for each patient to be analyzed against each other.
The time points in which interpolation was done was by determining the maximum shared time before data was collected on the subject. The fat fractions were also used to help model the pattern of the expected development of additional fat.

Since there are very few time points in between main data collection times, the interpolated data points are estimates, yet likely poor ones. Since we are referencing human data, there are many variables that can contribute to increasing the fat fraction. Additionally, only linear and quadratic interpolation techniques were used, thus resulting in more linear appearing data points.By interpolating data points in this way, it is assumed that the relationship is either linear or quadratic in nature.

Extrapolation was only done on a few subjects that had data points for the baseline and year 1, but lacked year 2 data. Extrapolation is much more unreliable as we only have means to model the current behavior and with extrapolation, you are predicting future behavior based on past known behavior. In interpolation, it is known that the data point will lie in between the two data points. For extrapolation, this is not the case; there will only be one data point that will be close to the extrapolated point that says very little about how the future results will present.

The main limitations have to do with the few time points collected. Since there are at maximum three times in which data was collected, it is difficult to gain sufficient data to model the progression of an individual subject, escpecially since FSHD develops over many years. Additionally, some subjects did not have data points for some measures and therefore extrapolation had to occur or subjects were excluded from the dataset.

### 1. Setup

```python
#import important libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from scipy.interpolate import interp1d

#reads the dataset when prompted for specific information
df = pd.read_csv("Module2_FSHD_Dataset.csv")
print(df.shape)
df.head()
```

### 2. Find Time Points to Interpolate at

```python
T1 = df['lag_months_y1'].min()   # earliest Year-1 visit across all subjects
T2 = df['lag_months_y2'].min()   # earliest Year-2 visit across all subjects
print(f"T1 = {T1:.2f} months")
print(f"T2 = {T2:.2f} months")
```

### 3. Example Interpolation of a Subjectc
#### i. Interpolation Calculations

```python
row = df.loc[df['subject_id'] == '13-0004'].iloc[0] ## picks example patient to interpolate

# creates arrays of time intervals between datapoints and fat fraction measures
lag_months = np.array([row['lag_months_baseline'], row['lag_months_y1'], row['lag_months_y2']])
ff_values = np.array([row['ff_baseline_R'], row['ff_y1_R'], row['ff_y2_R']])

print("lag_months:", lag_months)
print("ff_values: ", ff_values)

f_linear = interp1d(lag_months, ff_values, kind='linear') # computes linear interpolation between baseline and timepoint
f_quadratic = interp1d(lag_months, ff_values, kind='quadratic') # computes quadratic interpolation between first and second timepoint

print("f_linear(T1)   =", f_linear(T1))
print("f_quadratic(T2) =", f_quadratic(T2))
```

#### ii. Creation of Graph

```python
TEAL = "#1f7a8c"
PINK = "#e0569b"
ORANGE = "#e8821e"

x_curve = np.linspace(lag_months[0], lag_months[-1], 200)
quad_curve = f_quadratic(x_curve)

# plots linear and quadratic interpolations to estimate datapoints
plt.figure(figsize=(7.5, 5))
plt.plot(x_curve, quad_curve, color=ORANGE, lw=2.5, linestyle="--", label="Quadratic (all 3 points)")
plt.plot(lag_months[:2], ff_values[:2], color=PINK, lw=2.5, linestyle="--", label="Linear (baseline -> Year 1)")
plt.scatter(lag_months, ff_values, color=TEAL, s=90, zorder=4, label="Measured visits (known)")
plt.scatter([T1], [f_linear(T1)], color=PINK, marker="D", s=120, zorder=5, label=f"Linear estimate @ T1={T1:.1f}")
plt.scatter([T2], [f_quadratic(T2)], color=ORANGE, marker="D", s=120, zorder=5, label=f"Quadratic estimate @ T2={T2:.1f}")
plt.title("Subject 13-0004 — right tibialis anterior fat fraction")
plt.xlabel("Months since baseline (lag_months)")
plt.ylabel("Fat fraction (%)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
plt.savefig("Subject 13-0004 Interpolation")
```

### 4. Interpolation of all Patients

```python
# repeats interpolation process for every patient
results = []

for idx, row in df.iterrows():
    subject_id = row['subject_id']

    lag_months = np.array([row['lag_months_baseline'], row['lag_months_y1'], row['lag_months_y2']])
    ff_values = np.array([row['ff_baseline_L'], row['ff_y1_L'], row['ff_y2_L']])
    d4z4 = np.array(row['d4z4'])
    dux4 = np.array(row['dux4_signature_baseline_L'])

    if not (np.isnan(lag_months[:2]).any() or np.isnan(ff_values[:2]).any()):
        est_T1 = float(interp1d(lag_months[:2], ff_values[:2], kind='linear')(T1))
    else:
        est_T1 = np.nan

    if not (np.isnan(lag_months).any() or np.isnan(ff_values).any()):
        est_T2 = float(interp1d(lag_months, ff_values, kind='quadratic')(T2))
    else:
        est_T2 = np.nan

    results.append({'subject_id': subject_id,
                'estimate_T1_linear': est_T1,
                'estimate_T2_quadratic': est_T2, 'd4z4': d4z4, 'dux4_signature': dux4})
    pass


results_df = pd.DataFrame(results)
results_df.head()

results_df.to_csv(
    "interpolated_estimates.csv",
    index=False
)

print(results_df)
```

### 5. Utilizing of Interpolation Data

```python
percent_change = []

for idx, row in results_df.iterrows():
    if pd.notna(row['estimate_T1_linear']) and pd.notna(row['estimate_T2_quadratic']):
        change = (
            (row['estimate_T2_quadratic'] - row['estimate_T1_linear'])
            / row['estimate_T1_linear']
        ) * 100

        percent_change.append(change)

d4z4_list = []

for idx, row in results_df.iterrows():
    if pd.notna(row['estimate_T1_linear']) and pd.notna(row['estimate_T2_quadratic']):

        d4z4_list.append(int(row['d4z4']))

plt.scatter (d4z4_list, percent_change, color='blue')

dux4_list = []
for idx, row in results_df.iterrows():
    if pd.notna(row['estimate_T1_linear']) and pd.notna(row['estimate_T2_quadratic']) and pd.notna(row['dux4_signature']):
        dux4_list.append(row['dux4_signature'])

percent_change_for_dux4 = []

for idx, row in results_df.iterrows():
    if pd.notna(row['estimate_T1_linear']) and pd.notna(row['estimate_T2_quadratic']) and pd.notna(row['dux4_signature']):
        change = (
            (row['estimate_T2_quadratic'] - row['estimate_T1_linear'])
            / row['estimate_T1_linear']
        ) * 100

        percent_change_for_dux4.append(change)
```

### 6. Graphing Data

```python
plt.scatter (d4z4_list, percent_change, color='blue')
plt.xlabel('D4Z4 Repeats')
plt.ylabel('Percent Change in Left Leg Fat Fraction')
plt.title('Scatter Plot of D4Z4 vs. Change in Fat Fraction')

model = LinearRegression()
x_reg = np.array(d4z4_list).reshape(-1, 1)
y = percent_change

model.fit(x_reg, y)
slope = model.coef_[0]
intercept = model.intercept_
r2 = model.score(x_reg, y)
equation = f"y = {slope:.2f}x + {intercept:.2f}\nR^2 = {r2:.2f}"

plt.text(6, -20, equation, color = "blue", fontsize = 12, verticalalignment = "top")
plt.plot(x_reg, model.predict(x_reg), color = "blue")

plt.show()
plt.savefig("D4Z4 vs. Change in Fat Fraction.png")


plt.scatter (dux4_list, percent_change_for_dux4, color = 'blue')
plt.xlabel('Left Leg DUX4 Signature Baseline')
plt.ylabel('Percent Change in Left Leg Fat Fraction')
plt.title('Scatter Plot of DUX4 Signature vs. Change in Fat Fraction')

x_reg = np.array(dux4_list).reshape(-1, 1)
y = percent_change_for_dux4

model.fit(x_reg, y)
slope = model.coef_[0]
intercept = model.intercept_
r2 = model.score(x_reg, y)
equation = f"y = {slope:.2f}x + {intercept:.2f}\nR^2 = {r2:.2f}"

plt.text(3, -20, equation, color = "blue", fontsize = 12, verticalalignment = "top")
plt.plot(x_reg, model.predict(x_reg), color = "blue")

plt.show()
plt.savefig("DUX4 Signature vs. Change in Fat Fraction.png")
```

```python
#creates d4z4 groups based on the viable patients list to be used in the dotplot
d4z4_groups = sorted({patient.get_d4z4() for patient in viable_patients})
group_positions = {group: index for index, group in enumerate(d4z4_groups)}
group_counts = {group: 0 for group in d4z4_groups}
group_sizes = {
    group: sum(patient.get_d4z4() == group for patient in viable_patients)
    for group in d4z4_groups
}
dot_x = []
#adds a patient into the d4z4 groups depending on the result of the d4z4 getter Patient class method
for patient in viable_patients:
    group = patient.get_d4z4()
    within_group_index = group_counts[group]
    dot_x.append(
        group_positions[group]
        + (within_group_index - (group_sizes[group] - 1) / 2) * 0.08
    )
    group_counts[group] += 1
```

#### 5. Creating the Graph

```python
#settings related to the dotplot made from the d4z4 and year 2 to baseline fat fraction difference
fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(dot_x, net_ff_diff, color="darkorange", edgecolor="black", alpha=0.8)
ax.axhline(0, color="black", linewidth=0.8)
group_changes = {
    group: [
        change
        for patient, change in zip(viable_patients, net_ff_diff)
        if patient.get_d4z4() == group
    ]
    for group in d4z4_groups
}
#finds the means of the patients within the groups based on d4z4
group_means = {
    group: statistics.mean(changes)
    for group, changes in group_changes.items()
}
#graphs the data points
ax.scatter(
    [group_positions[group] for group in d4z4_groups],
    [group_means[group] for group in d4z4_groups],
    marker="D",
    color="black",
    s=35,
    label="Group mean",
    zorder=3,
)
#calculating the standard deviations
groups_with_sd = [
    group for group in d4z4_groups if len(group_changes[group]) > 1
]
#adds errorbars based on standard deviations
ax.errorbar(
    [group_positions[group] for group in groups_with_sd],
    [group_means[group] for group in groups_with_sd],
    yerr=[
        statistics.stdev(group_changes[group])
        for group in groups_with_sd
    ],
    fmt="none",
    ecolor="black",
    capsize=5,
    linewidth=1.5,
    label="±1 SD",
    zorder=2,
)
#titles and axises added and plot generated
ax.set_xticks(range(len(d4z4_groups)), d4z4_groups)
ax.set_xlabel("D4Z4 group")
ax.set_ylabel("Total fat fraction change from baseline (%)")
ax.set_title("Fat Fraction Increase by D4Z4 Group")
ax.grid(axis="y", alpha=0.3)
ax.legend()
fig.tight_layout()
plt.show()
plt.savefig("d4z4 Versus Increase in Right Leg Fat Fraction from Baseline to Year 2")
```
